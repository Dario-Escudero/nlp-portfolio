## Section 2 — Plot Styles, Colors & Markers

Your lecture showed format strings like `'b+'` (blue plus), `'r--'` (red dashed line), `'g^'` (green triangles).

The format string is: `'[color][marker][linestyle]'`

| Color | Marker | Linestyle |
|-------|--------|-----------|
| `b` blue | `+` plus | `-` solid |
| `r` red | `o` circle | `--` dashed |
| `g` green | `^` triangle | `:` dotted |
| `k` black | `s` square | `-.` dash-dot |

````python
# DEMO — three curves on the same plot
x = np.linspace(0, 5, 40)

plt.plot(x, x,    'r--', label='Linear')   # red dashed
plt.plot(x, x**2, 'b+',  label='Square')   # blue plus
plt.plot(x, x**3, 'g^',  label='Cubic')    # green triangle

plt.legend()   # shows the labels
plt.title('Three curves on one plot')
plt.show()
````

````python
# SOLUTION 2
x = np.linspace(0, 2*np.pi, 100)

plt.plot(x, np.sin(x),  'k-',  label='sin(x)')
plt.plot(x, np.cos(x),  'r--', label='cos(x)')
plt.plot(x, -np.sin(x), 'go',  label='-sin(x)')

plt.xlabel('radians')
plt.title('Trig functions')
plt.legend()
plt.show()
````

## Section 3 — Subplots

Your lecture covered `plt.subplot(rows, cols, index)`.  
The index counts left→right, top→bottom, starting at 1.

```
subplot(2, 2, 1) | subplot(2, 2, 2)
─────────────────────────────────────
subplot(2, 2, 3) | subplot(2, 2, 4)
```

````python
# DEMO — 2x2 subplot grid
x = np.linspace(0, 5, 50)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(x, x)
plt.title('Linear')

plt.subplot(2, 2, 2)
plt.plot(x, x**2)
plt.title('Square')

plt.subplot(2, 2, 3)
plt.plot(x, x**3)
plt.title('Cubic')

plt.subplot(2, 2, 4)
plt.plot(x, np.exp(x))
plt.title('Exponential')

plt.tight_layout()   # prevents overlapping titles
plt.show()
````

## Section 4 — Histogram

Histograms show how data is **distributed**.  
`plt.hist(data, bins=N)` — `bins` controls how many bars you see.

````python
# DEMO — histogram of random normal data
data = np.random.randn(1000)   # 1000 values from a normal distribution

plt.hist(data, bins=30, color='steelblue', edgecolor='white')
plt.xlabel('Value')
plt.ylabel('Count')
plt.title('Histogram of 1000 random normal values')
plt.show()
````

## Section 5 — Box Plot

Box plots show **median, quartiles, and outliers** in one compact visual.  
`plt.boxplot(data)` — or pass a list of arrays to compare groups side by side.

````python
# DEMO — boxplot comparing three distributions
group_a = np.random.normal(loc=50,  scale=10, size=200)
group_b = np.random.normal(loc=60,  scale=20, size=200)
group_c = np.random.exponential(scale=15, size=200)  # skewed, will have outliers

plt.boxplot([group_a, group_b, group_c], labels=['Group A', 'Group B', 'Group C (skewed)'])
plt.ylabel('Value')
plt.title('Box Plot — three groups')
plt.show()
````

## Section 6 — Scatter Plot

Scatter plots reveal **relationships between two numeric variables**.  
`plt.scatter(x, y, c=colors, s=sizes, alpha=0.5)`

````python
# DEMO
sales  = np.random.exponential(scale=500, size=300)
profit = sales * np.random.uniform(0.05, 0.3, 300) - np.random.uniform(0, 50, 300)

plt.scatter(sales, profit, alpha=0.4, color='steelblue')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sales vs Profit')
plt.show()
````

## 🏆 Final Challenge — Put It All Together

This exercise ties everything together in one dashboard-style figure.

````python
# ✏️ FINAL CHALLENGE
#
# Generate a fake e-commerce dataset:
#   n = 500 orders
#   sales    = np.random.exponential(scale=300, size=n)
#   discount = np.random.uniform(0, 0.5, size=n)
#   profit   = sales * (0.2 - discount) + np.random.randn(n)*20
#   category = np.random.choice(['Electronics','Books','Food','Clothing'], size=n)
#
# Create a 2x2 figure with:
#   [1,1] Histogram of sales (log y-scale)
#   [1,2] Boxplot of profit grouped by category (one box per category)
#   [2,1] Scatter plot: sales vs profit, colored by discount amount
#   [2,2] Bar chart: mean sales per category
#
# Tip for boxplot by category:
#   groups = [profit[category==c] for c in ['Electronics','Books','Food','Clothing']]
#   plt.boxplot(groups, labels=[...])

# TODO: your code here
````

````python
# SOLUTION — Final Challenge
np.random.seed(42)
n = 500
sales    = np.random.exponential(scale=300, size=n)
discount = np.random.uniform(0, 0.5, size=n)
profit   = sales * (0.2 - discount) + np.random.randn(n)*20
cats     = np.array(['Electronics','Books','Food','Clothing'])
category = np.random.choice(cats, size=n)

plt.figure(figsize=(13, 9))

# [1,1] Histogram
plt.subplot(2, 2, 1)
plt.hist(sales, bins=40, color='steelblue', edgecolor='white')
plt.yscale('log')
plt.xlabel('Sales'); plt.ylabel('Count (log)')
plt.title('Sales Distribution (log scale)')

# [1,2] Boxplot by category
plt.subplot(2, 2, 2)
groups = [profit[category==c] for c in cats]
plt.boxplot(groups, labels=cats)
plt.ylabel('Profit'); plt.title('Profit by Category')
plt.xticks(rotation=15)

# [2,1] Scatter sales vs profit colored by discount
plt.subplot(2, 2, 3)
sc = plt.scatter(sales, profit, c=discount, cmap='RdYlGn_r', alpha=0.5, s=20)
plt.colorbar(sc, label='Discount rate')
plt.xlabel('Sales'); plt.ylabel('Profit')
plt.title('Sales vs Profit (color = discount)')

# [2,2] Bar chart mean sales per category
plt.subplot(2, 2, 4)
mean_sales = [sales[category==c].mean() for c in cats]
plt.bar(cats, mean_sales, color=['#4e79a7','#f28e2b','#59a14f','#e15759'])
plt.ylabel('Mean Sales ($)'); plt.title('Average Sales per Category')
plt.xticks(rotation=15)

plt.suptitle('E-Commerce Dashboard', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````
